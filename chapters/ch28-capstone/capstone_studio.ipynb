## B. Fit, freeze, and reconstruct

The baseline uses mean squared adjacent differences, then the midpoint between development class-mean features. It returns a threshold decision, not a disease probability. Inspect fitting IDs. Reconstruct one evaluation record's feature and rule without changing the model. Does a matching artifact digest establish its origin or scientific meaning?


```python
model = studio.fit_baseline(fixture)
held_out = next(r for r in fixture["records"] if r["partition"] == "evaluation")
value = studio.feature(held_out)
print("Fitting people:", model["fit_people"])
print("Fitting record IDs:", model["fit_ids"])
print("Model ID:", model["model_id"])
print("Feature / threshold / reconstructed prediction:", value, model["threshold"], int(value > model["threshold"]))
print(json.dumps(studio.infer(held_out, model), indent=2))
```

Accuracy is conditional on accepted records. Coverage uses all intended records; people and record counts remain visible. Development is a fitted-data description. ## F. Change a copy and explain the comparison

The digest exercise detects a changed file relative to the saved receipt. It does not show that an attacker cannot replace both. The added note is an intentional byte change, not a claim about your observed accuracy. Preserve the original directory and submit both evidence sets with a brief explanation.


```python
modified = folder.with_name(folder.name + "_modified_" + uuid.uuid4().hex[:6])
shutil.copytree(folder, modified)
report_path = modified / "report.json"
changed = json.loads(report_path.read_text(encoding="utf-8"))
changed["student_note"] = "Intentional edit for receipt comparison; original bundle preserved."
report_path.write_text(json.dumps(changed, indent=2) + "\n", encoding="utf-8")
print("Original comparison:", studio.verify_receipt(folder))
print("Modified comparison:", studio.verify_receipt(modified))
```
