# Chapter 2: From neuronal events to measurable signals

Simulate oscillations, transients, and additive sensor noise to distinguish cellular activity from macroscopic field recordings.

**Education only, not clinical software.** Run cells in order and explain results and limitations.

```python
import numpy as np, matplotlib.pyplot as plt
rng = np.random.default_rng(42)
fs=256; t=np.arange(0,4,1/fs)
alpha=12*np.sin(2*np.pi*10*t)
slow=7*np.sin(2*np.pi*2*t)
transient=35*np.exp(-0.5*((t-2)/0.045)**2)
noise=rng.normal(0,5,len(t))
mixture=alpha+slow+transient+noise
fig,ax=plt.subplots(2,1,figsize=(11,6),sharex=True)
ax[0].plot(t,alpha,label="Illustrative 10-Hz oscillation")
ax[0].plot(t,slow,label="Illustrative 2-Hz oscillation")
ax[0].legend()
ax[1].plot(t,mixture,label="Synthetic sensor-level mixture")
ax[1].set(xlabel="Time (s)",ylabel="Arbitrary amplitude");ax[1].legend()
plt.tight_layout();plt.show()
print("Synthetic teaching model: NOT a forward model, neuronal spike recording, or patient EEG.")
```
