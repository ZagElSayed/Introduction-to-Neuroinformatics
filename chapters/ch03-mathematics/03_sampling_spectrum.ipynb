# Chapter 3: Sampling, aliasing and Fourier spectra

Investigate a synthetic 35-Hz signal sampled above and below the Nyquist limit.

**Education only, not clinical software.** Run cells in order and explain results and limitations.

```python
import numpy as np, matplotlib.pyplot as plt
f=35
fig,ax=plt.subplots(2,1,figsize=(10,6))
for i,fs in enumerate([200,50]):
    t=np.arange(0,1,1/fs)
    x=np.sin(2*np.pi*f*t)
    spectrum=np.abs(np.fft.rfft(x))
    freqs=np.fft.rfftfreq(len(x),1/fs)
    ax[i].stem(freqs,spectrum,basefmt=" ")
    ax[i].set(xlabel="Frequency (Hz)",ylabel="FFT magnitude",title=f"Sampling {f} Hz at {fs} Hz")
plt.tight_layout();plt.show()
print("At 50 Hz sampling, 35 Hz aliases to 15 Hz. Antialias filters must act before downsampling.")
```
