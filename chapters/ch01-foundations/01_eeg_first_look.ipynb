# Chapter 1: First look at a real EEG

Load PhysioNet S001R01 EDF, inspect metadata and plot 10 seconds.

**Education only, not clinical software.** Run cells in order and explain results and limitations.

```python
from pathlib import Path
import mne, numpy as np, matplotlib.pyplot as plt
edf = Path("S001R01.edf")  # Download from PhysioNet; keep locally
assert edf.exists(), "Download S001R01.edf from the PhysioNet EEGMMIDB dataset first."
raw = mne.io.read_raw_edf(str(edf), preload=False)
print("Channels:", len(raw.ch_names), "Sampling rate:", raw.info["sfreq"], "Hz")
print("Duration (s):", raw.n_times / raw.info["sfreq"])
picks = mne.pick_types(raw.info, eeg=True)[:5]
data, times = raw[picks, :int(10 * raw.info["sfreq"])]
plt.figure(figsize=(12,5))
for k, trace in enumerate(data):
    plt.plot(times, trace * 1e6 + 100*k, label=raw.ch_names[picks[k]])
plt.xlabel("Time (s)"); plt.ylabel("Amplitude (µV), vertically offset")
plt.legend(); plt.show()
```
