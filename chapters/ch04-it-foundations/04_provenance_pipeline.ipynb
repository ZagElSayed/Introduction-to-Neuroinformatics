Build a small provenance record for a synthetic recording, with a checksum, units, parameters, and derived summary.

**Education only, not clinical software.** Run cells in order and explain results and limitations.

```python
import numpy as np, json, hashlib, platform
from pathlib import Path
rng=np.random.default_rng(17); fs=256
signal=rng.normal(size=fs*5).astype("<f8")
raw_bytes=signal.tobytes()
digest=hashlib.sha256(raw_bytes).hexdigest()
out=Path("outputs/ch04");out.mkdir(parents=True,exist_ok=True)
np.save(out/"synthetic_signal.npy",signal)
summary={"source":"synthetic, not clinical","sampling_rate_hz":fs,"units":"arbitrary",
"seed":17,"source_array_sha256":digest,"samples":len(signal),
"rms":float(np.sqrt(np.mean(signal**2))),"python_version":platform.python_version(),
"numpy_version":np.__version__}
(out/"provenance.json").write_text(json.dumps(summary,indent=2))
assert hashlib.sha256(np.load(out/"synthetic_signal.npy").tobytes()).hexdigest()==digest
print(json.dumps(summary,indent=2))
```
